# Recommendation Baselines
Minh họa popularity, content-based và hybrid recommendation.

```python
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

papers=pd.DataFrame([
 {'id':1,'text':'retrieval augmented generation evidence','citations':18000},
 {'id':2,'text':'sentence embedding semantic search','citations':20000},
 {'id':3,'text':'citation graph pagerank network','citations':30000},
 {'id':4,'text':'marine biology fish ecosystem','citations':500},
])
saved_ids=[1,2]
X=TfidfVectorizer().fit_transform(papers.text)
profile=X[papers.id.isin(saved_ids)].mean(axis=0)
content=cosine_similarity(profile,X).ravel()
popularity=np.log1p(papers.citations)/np.log1p(papers.citations.max())
hybrid=.65*content+.35*popularity
papers.assign(content=content,popularity=popularity,hybrid=hybrid).query('id not in @saved_ids').sort_values('hybrid',ascending=False)
```

```python
def precision_at_k(recommended, relevant, k):
    return len(set(recommended[:k]) & set(relevant))/k

recommended=papers.assign(score=hybrid).query('id not in @saved_ids').sort_values('score',ascending=False).id.tolist()
precision_at_k(recommended, relevant=[3], k=2)
```
