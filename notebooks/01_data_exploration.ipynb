# Data Exploration
Khám phá corpus seed: phân bố năm, citation, Open Access và dữ liệu thiếu.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

root = Path('..') if Path.cwd().name == 'notebooks' else Path('.')
rows = [
    {'title': 'Attention Is All You Need', 'year': 2017, 'citations': 130000, 'open_access': True, 'topic': 'NLP'},
    {'title': 'BERT', 'year': 2018, 'citations': 90000, 'open_access': True, 'topic': 'NLP'},
    {'title': 'RAG', 'year': 2020, 'citations': 18000, 'open_access': True, 'topic': 'IR'},
    {'title': 'Graph Neural Networks Review', 'year': 2021, 'citations': 6000, 'open_access': False, 'topic': 'Graph'},
]
df = pd.DataFrame(rows)
df
```

```python
df.describe(include='all')
```

```python
ax = df.groupby('year')['citations'].sum().plot(kind='bar', title='Citations by publication year')
ax.set_ylabel('Citations')
plt.show()
```

```python
missing = df.isna().mean().sort_values(ascending=False)
missing.to_frame('missing_rate')
```

Citation có phân bố lệch phải; khi huấn luyện/ranking nên dùng `log1p(citations)` thay vì giá trị thô.
