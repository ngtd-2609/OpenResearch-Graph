# Neural Relevance Model Evaluation
Đọc dataset query–title–abstract, train mô hình nhỏ và phân tích lỗi.

```python
from pathlib import Path
import json
import pandas as pd

root=Path('..') if Path.cwd().name=='notebooks' else Path('.')
data_path=root/'backend/data/relevance_sample.json'
rows=json.loads(data_path.read_text(encoding='utf-8'))
df=pd.DataFrame(rows)
df.groupby('label').size().rename('count')
```

```python
df.assign(query_length=df['query'].str.split().str.len(), document_length=(df['title']+' '+df['abstract']).str.split().str.len()).groupby('label')[['query_length','document_length']].mean()
```

```python
# Chạy từ thư mục backend để train thật:
# python -m app.ml.training.train_relevance --config configs/relevance.yaml
# Checkpoint lưu model config, training config và validation metrics.
```

```python
from sklearn.metrics import confusion_matrix, classification_report
# Thay bằng output model thật sau khi train. Ví dụ minh họa cấu trúc error analysis:
y_true=[1,0,1,0,1,0]
y_pred=[1,0,1,1,0,0]
print(confusion_matrix(y_true,y_pred))
print(classification_report(y_true,y_pred,zero_division=0))
```
